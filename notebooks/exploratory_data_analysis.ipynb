```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/processed_housing_data.csv')
```

```python
# Correlation Matrix
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
```

```python
# Scatter Plot of sqft_living vs list_price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='sqft_living', y='list_price', hue='property_type_SINGLE_FAMILY')
```

```python
# Scatter Plot of list_price by price_per_sqft
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='price_per_sqft', y='list_price')
```

```python
# Scatter Plot of list_price by days_on_market
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='days_on_market', y='list_price')
```

```python
# Scatter Plot of list_price by hoa_fee
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hoa_fee', y='list_price')
```

```python
# Scatter Plot of list_price by latitude
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='latitude', y='list_price')
```

```python
# Box Plot of list_price by zip_code
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='zip_code', y='list_price')
```

```python
# scatter plot of list_price by house_age
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='house_age', y='list_price')
```
